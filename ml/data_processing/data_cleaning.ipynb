```python
import pandas as pd
import re
```

```python
df = pd.read_csv("../data/latex_equation_groundtruth.csv")
df = df[["equation", "image_path"]]
df.shape
```

```python
df = df.dropna()
df.shape
```

```python
def is_valid_equation (equation):
    allowed_commands = ["\\left", "\\right", "\\frac", "\\cdot"]
    commands = re.findall(r"\\[a-zA-Z]+", equation)
    for command in commands:
        if command not in allowed_commands:
            return False
        
    # Remove commands from the equation
    cleaned_equation = re.sub(r"\\[a-zA-Z]+", "", equation)
    
    # check for at least one digit
    if not re.search(r"\d", cleaned_equation):
        return False
    
    # Final check for allowed characters
    digit_equation = re.sub(r"[+=(){} -]", "", cleaned_equation)
    if re.search(r"[^0-9.]", digit_equation):
        return False
    
    return True

df = df[df["equation"].apply(is_valid_equation)]
selected_df = df[["image_path", "equation"]]
selected_df.to_csv("../data/selected_images.csv", index=False)
selected_df.shape
```
